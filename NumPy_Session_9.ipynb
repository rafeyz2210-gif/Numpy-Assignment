## Task 1 — Create the Prices Array

### Assignment question
Create a NumPy array named `prices` with these values: `[299, 499, 799, 0, 1599, -1, 899]`. Print the array and its data type.

### Simple explanation
This creates the exact starting price array for the case study.

```python
prices = np.array([299, 499, 799, 0, 1599, -1, 899])

print("Prices:", prices)
print("Data type:", prices.dtype)
```

## Task 2 — Clean Invalid Prices

### Assignment question
Some entries in the prices array are invalid (0 or negative). Replace all values less than or equal to zero with the average of the remaining positive prices. Use boolean indexing and the `mean()` function.

### Simple explanation
Boolean indexing selects the positive prices. Their mean is used to replace invalid values.

```python
positive_prices = prices[prices > 0]
positive_mean = positive_prices.mean()

cleaned_prices = prices.copy()
cleaned_prices[cleaned_prices <= 0] = positive_mean

print("Average of positive prices:", positive_mean)
print("Cleaned prices:", cleaned_prices)
```

## Task 3 — Calculate Item Bills

### Assignment question
Suppose you have a NumPy array `quantities = [2, 1, 3, 4, 2, 1, 5]`. Calculate the total bill for each item by multiplying the cleaned prices array with quantities, and print the resulting array.

### Simple explanation
Multiplication is element-wise: each cleaned price is multiplied by the quantity at the same position.

```python
quantities = np.array([2, 1, 3, 4, 2, 1, 5])

item_bills = cleaned_prices * quantities

print("Quantities:", quantities)
print("Bill for each item:", item_bills)
```

## Task 4 — Summary Report

### Assignment question
Generate and print a summary report showing the minimum, maximum, average, and total of the cleaned prices array, and also the total bill for all items combined. Use NumPy functions like `min()`, `max()`, and `sum()`.

### Simple explanation
These aggregate functions reduce the array to summary values.

```python
print("Minimum price:", cleaned_prices.min())
print("Maximum price:", cleaned_prices.max())
print("Average price:", cleaned_prices.mean())
print("Total of cleaned prices:", cleaned_prices.sum())
print("Total bill for all items:", item_bills.sum())
```

## Task 5 — Count Unique Price Values

### Assignment question
Use a NumPy function or method to find how many unique price values are present in the cleaned prices array. Print the suggested method and the result.

### Simple explanation
`np.unique()` returns distinct values. `len()` counts how many distinct values there are.

```python
unique_prices = np.unique(cleaned_prices)

print("Method: np.unique()")
print("Unique price values:", unique_prices)
print("Number of unique price values:", len(unique_prices))
```
