## Task 1 — Combine Restaurant Ratings

### Assignment question
Create two NumPy arrays representing the ratings of 5 restaurants on Zomato and Swiggy, then use `np.concatenate()` to combine them into a single array of 10 ratings and print the result.

### Simple explanation
`np.concatenate()` joins the two arrays together.

```python
import numpy as np

zomato_ratings = np.array([4.2, 3.8, 4.5, 4.0, 4.3])
swiggy_ratings = np.array([4.1, 3.9, 4.4, 4.2, 4.0])

combined_ratings = np.concatenate((zomato_ratings, swiggy_ratings))

print("Zomato ratings:", zomato_ratings)
print("Swiggy ratings:", swiggy_ratings)
print("Combined ratings:", combined_ratings)
```

## Task 2 — Stack Weekly Instagram Likes

### Assignment question
Given three arrays representing the number of likes on three different Instagram posts over 7 days, stack them vertically using `np.vstack()` so that each row represents one post's weekly likes, and print the stacked array.

### Simple explanation
`np.vstack()` stacks the arrays vertically, making each post one row.

```python
post1 = np.array([120, 150, 180, 160, 200, 220, 250])
post2 = np.array([90, 110, 130, 125, 150, 170, 190])
post3 = np.array([200, 210, 230, 240, 260, 280, 300])

stacked_likes = np.vstack((post1, post2, post3))

print("Stacked weekly likes:")
print(stacked_likes)
```

## Task 3 — Split Flipkart Product IDs

### Assignment question
You have an array of 12 Flipkart product IDs. Use `np.array_split()` to divide this array into 5 nearly equal parts, display each part, and check the shape of each split to confirm the division.

### Simple explanation
`np.array_split()` divides the array into nearly equal parts, even when an exact equal split is impossible.

```python
product_ids = np.array([101, 102, 103, 104, 105, 106, 107, 108, 109, 110, 111, 112])

parts = np.array_split(product_ids, 5)

for i, part in enumerate(parts, start=1):
    print(f"Part {i}:", part)
    print("Shape:", part.shape)
    print()
```

## Task 4 — WhatsApp Group Chat

### Assignment question
Simulate a WhatsApp group chat: create a 2D NumPy array where each row is a user and each column is the number of messages sent per day for a week. Use `np.insert()` to add a new user (row) with their message counts, then use `np.delete()` to remove the user who sent the least messages overall.

### Simple explanation
`np.insert()` adds the new row. `sum(axis=1)` finds each user's weekly total, and `np.delete()` removes the row with the smallest total.

```python
messages = np.array([
    [12, 15, 10, 8, 14, 11, 9],
    [20, 18, 22, 16, 19, 21, 17],
    [5, 7, 4, 6, 5, 8, 3]
])

new_user = np.array([10, 12, 9, 11, 13, 10, 14])
messages_with_new_user = np.insert(messages, 3, new_user, axis=0)

weekly_totals = messages_with_new_user.sum(axis=1)
least_user_index = np.argmin(weekly_totals)

final_messages = np.delete(messages_with_new_user, least_user_index, axis=0)

print("Original messages:")
print(messages)
print("
After adding new user:")
print(messages_with_new_user)
print("
Weekly totals:")
print(weekly_totals)
print("
Index of user with least messages:", least_user_index)
print("
After removing that user:")
print(final_messages)
```

## Task 5 — `view()` vs `copy()`

### Assignment question
Given a NumPy array of YouTube video view counts, use `.view()` to create a view and `.copy()` to create a copy. Modify the first element in each and print all arrays to demonstrate the difference between view and copy.

### Simple explanation
A view shares the original data, while a copy has independent data. Therefore changing the view changes the original, but changing the copy does not.

```python
views = np.array([1000, 2500, 1800, 3200, 4100])

views_view = views.view()
views_copy = views.copy()

views_view[0] = 9999
views_copy[0] = 7777

print("Original array:", views)
print("View:", views_view)
print("Copy:", views_copy)
```
