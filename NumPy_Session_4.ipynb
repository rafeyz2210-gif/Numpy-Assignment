```python
import numpy as np

```

```python
likes = np.array([120,250,180,450,320,150,500])
comments = np.array([10,20,15,35,25,12,40])

engagement = likes + comments

print("engagement per post", engagement)
```

```python
prices = np.array([200,150,300,250,180])
discounts = np.array([20,15,50,25,30])

final_prices = prices - discounts

print("final prices", final_prices)
```

```python
scores = np.array([175,210,165,195,180])

result = scores > 180
print("scores:",scores)
print("scores greater than 180:",result)
```

```python
paytm = np.array([1,0,1,0,0,1])
phonepe = np.array([0,1,0,0,1,0])

result = np.logical_or(paytm == 1, phonepe ==1)

print("paytm:", paytm)
print("phonepe:",phonepe)
print("paid through either app:",result)
```
