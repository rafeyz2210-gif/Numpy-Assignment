## Task 1 — Unique IPL Team Names

### Assignment question
Given a NumPy array of IPL team names with some duplicates, use `np.unique()` to print a sorted list of all unique team names.

### Simple explanation
`np.unique()` removes duplicates and returns the unique values in sorted order.

```python
teams = np.array(["MI", "CSK", "RCB", "MI", "GT", "CSK", "KKR", "RCB", "GT", "MI"])

unique_teams = np.unique(teams)

print("Unique IPL teams:", unique_teams)
```

## Task 2 — Missing Zomato Ratings

### Assignment question
Create a NumPy array of Zomato order ratings with some NaN values, then use `np.isnan()` to count how many ratings are missing.

### Simple explanation
`np.isnan()` gives True for NaN values. Adding the True values counts the missing ratings.

```python
ratings = np.array([4.5, 4.0, np.nan, 3.5, np.nan, 4.2, 5.0, np.nan])

missing = np.isnan(ratings)

print("Ratings:", ratings)
print("Missing-value mask:", missing)
print("Number of missing ratings:", np.sum(missing))
```

## Task 3 — Clip Flipkart Prices

### Assignment question
Given an array of Flipkart product prices, use `np.clip()` to limit all prices between 100 and 1000, and print the resulting array.

### Simple explanation
`np.clip()` changes values below 100 to 100 and values above 1000 to 1000.

```python
prices = np.array([50, 120, 450, 999, 1200, 75, 1500, 700])

clipped_prices = np.clip(prices, 100, 1000)

print("Original prices:", prices)
print("Clipped prices:", clipped_prices)
```

## Task 4 — Handle NaN and Infinity

### Assignment question
You have a NumPy array of YouTube video view counts, some of which are NaN or inf. Replace all NaN values with 0, and all inf values with the maximum finite value in the array.

### Simple explanation
Find the maximum finite value first. Then replace NaN with 0 and positive infinity with that maximum.

```python
views = np.array([1200.0, np.nan, 3500.0, np.inf, 2200.0, np.nan, 4800.0])

finite_values = views[np.isfinite(views)]
max_finite = np.max(finite_values)

cleaned_views = views.copy()
cleaned_views[np.isnan(cleaned_views)] = 0
cleaned_views[np.isposinf(cleaned_views)] = max_finite

print("Original views:", views)
print("Maximum finite value:", max_finite)
print("Cleaned views:", cleaned_views)
```

## Task 5 — Find Indices of Even Numbers

### Assignment question
Use Python/NumPy to find the indices of all even numbers in a NumPy array using `np.where()`, then test the code with your own example array.

### Simple explanation
`np.where()` returns the positions where the condition is True. `numbers % 2 == 0` identifies even numbers.

```python
numbers = np.array([11, 24, 35, 42, 57, 68, 73, 80])

even_indices = np.where(numbers % 2 == 0)

print("Array:", numbers)
print("Indices of even numbers:", even_indices)
print("Even numbers:", numbers[even_indices])
```
