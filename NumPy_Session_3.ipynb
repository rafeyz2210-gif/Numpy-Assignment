```python
import numpy as np
```

```python
likes = np.array([120,250,180,450,320,150,500])

print("array:", likes)
print("ndim:", likes.ndim)
print("shape:", likes.shape)
print("size:",likes.size)
print("dtype:",likes.dtype)
print("itemsize:",likes.itemsize)
print("nbytes:",likes.nbytes)
```

```python
steps = np.array([
    [3200,4500],
    [2800,5100],
    [4000,4800],
    [3500,5200],
    [3000,4700]
])

print("original array:")
print(steps)
print("shape:",steps.shape)
```

```python
steps_id = steps.reshape(10)

print("1d array:")
print(steps_id)
print("shape:",steps_id.shape)
```

```python
steps_2d = steps_id.reshape(5,2)

print("back to 2d:")
print(steps_2d)
print("shape:",steps_2d.shape)
```

```python
prices = np.array([
    120,80,150,200,90,180,75,250,110,160,95,220
])

print("original array:")
print(prices)

```

```python
prices_2d = prices.reshape(3,4)

print("3x4 array:")
print(prices_2d)
```

```python
prices_ravel = prices_2d.ravel()

print("ravel:")
print(prices_ravel)
```

```python
prices_flatten = prices_2d.flatten()

print("flatten:")
print(prices_flatten)
```

```python
prices_resize = prices_2d.copy()
prices_resize.resize(4,3)

print("resize:")
print(prices_resize)
```

```python
spotify = np.array([
    [12,8,5],
    [7,15,9,],
    [10,6,14]
    ])

print("original array:")
print(spotify)
```

```python
spotify_T = spotify.T

print("Using .T:")
print(spotify_T)
```

```python
spotify_transpose = np.transpose(spotify)

print("Using np.transpose():")
print(spotify_transpose)
```
