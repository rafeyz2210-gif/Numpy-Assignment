```python
import numpy as np
```

```python
my_likes = np.array([120,250,180,450,320,150,500])
friend_likes = np.array([100,200,220,400,300,180,450])

combined = np.add(my_likes,friend_likes)
difference = np.subtract(my_likes,friend_likes)
print("combined likes:", combined)
print("differences:",difference)
```

```python
prices = np.array([200,150,300,250,180])

discounted_prices = np.multiply(prices,0.9)
total_bill = np.sum(discounted_prices)
print("original prices:",prices)
print("prices after 10% discount:",discounted_prices)
print("final bill:",total_bill)
```

```python
np.sum(discounted_prices)
```

```python
steps = np.array([5000,6500,7200,4800,8000,
                  6000,5500,7000,6200,7500,
                  4500,6800,7100,5900,8200,
                  5300,6400,7600,5800,6900,
                  4700,7300,6600,6100,7800,
                  5200,7000,6300,7400,5600
                  ])

print("mean:",np.mean(steps))
print("median:",np.median(steps))
print("standard deviation:",np.std(steps))
print("maximum:",np.max(steps))
```

```python
ratings = np.random.uniform(1,5,10)

print("original  ratings:")
print(ratings)
print("rounded:",np.round(ratings))
print("rounded down:",np.floor(ratings))
print("rounded up:",np.ceil(ratings))
```

```python
skips = np.array([
    2, 5, 1, 8, 3,
    6, 4, 7, 2, 9,
    5, 3, 6, 1, 4,
    8, 7, 2, 5, 6
])

percentile_75 = np.percentile(skips, 75)

print("75th percentile of skips:", percentile_75)
```
